# List
- Represented by big brackets []
- mutable
- heterogeneous (can contain mix of datatypes)

### Creating Lists

Two methods for creating a list:
- literals (using brackets)
- list() constructor

```python
numbers = [1,2,3,4,5]
```

### Accessing Items in the list

Indexing can be positive and negative. Positive from the start, negative from the last.

```python
print(numbers[-3])
```

### Methods
- .append()
- .remove()
- .sort()
- many more...

```python
numbers.append(6)
print(numbers)
```

```python
numbers.remove(3)
print(numbers)
```

```python
random = [1,'Ram',4.5]
print(random)
```

```python
numbers.sort()
print(numbers)
```

```python
unsorted = [5,4,9,8]
unsorted.sort()
print(unsorted)
```

### List Comprehension

Generating list based on the iterables.

```python
square = [num**2 for num in range(0,10)]
print(square)
```

# Dictionary

- Mutable collection of key value pairs
- contains unique keys.
- can contain list as values.

### Creating a dictionary

```python
colors = { 'rojo':"red",2:5,3:"green"}
```

```python
colors
```

### Accessing values from dictionary

```python
colors[2]
```

### Updating values

```python
colors[2] = 10
```

```python
colors
```

```python
random = {1:"red",2:"blue",3:["black","white"]}
```

```python
random
```

```python
random[3].remove('white')
```

```python
random
```

# Set

### Creating a set

```python
random = [1,2,3,3,3]
set(random)
```

### Pass string as string is iterable

The order is not preserved in set

```python
name = 'Harryh'
name_set = set(name)
```

```python
color = 'red'
color_set=set(color)
```

```python
#union operation
# print(name_set | color_set)
name_set.union(color_set)
```
