## Separating Columns

### Numerical Columns

- .mean(), .min(), .max(), .std(), .std(),  etc
- .corr() gives correlation between numerical variables
- .describe() gives mean, max,.. in numerical columns

```python
numerical_cols = df.select_dtypes(include=['float64', 'int64'])

print("Numerical Columns:")
print(numerical_cols.columns.tolist())
print("\n" + "="*60)

print("Describe:")
print(numerical_cols.describe())
print("\n" + "="*60)

print("Mean:")
print(numerical_cols.mean())
print("\n" + "="*60)

print("Correlation:")
print(numerical_cols.corr())
```

### Categorical Columns

.describe() gives count,unique,top,freq in categorical columns

```python
categorical_cols = df.select_dtypes(include=['object'])

print("Categorical Columns:")
print(categorical_cols.columns.tolist())
print("\n" + "="*60)

print(categorical_cols.describe())
```

## Applying functions

```python
# .assign() -> helps to create new column

#Add regional sales together
df = df.assign(Regional_Sales = df['NA_Sales'] + df['EU_Sales'])

df.head()
```

```python
# apply lambda function

# Simple Example: Label games as "Old" or "New"
df['Era'] = df['Year'].apply(lambda year: 'New' if year >= 2010 else 'Old')

df.head()
```

## Filling Null values

```python
df.isnull().sum()
```

### Fill Year

```python
df['Year'].describe()
```

```python
median_year = df['Year'].median()
print(median_year)
df['Year'] = df['Year'].fillna(median_year)
print(f"Missing year values: {df['Year'].isnull().sum()}")
```

### Fill Published By

```python
df['Published By'].describe()
```

```python
df['Published By'] = df['Published By'].fillna('Unknown Publisher')
print(f"Missing values: {df['Published By'].isnull().sum()}")
```

```python
print("Games with 'Unknown Publisher':")
print(df[df['Published By'] == 'Unknown Publisher'][['Name','Published By']].head())
```
