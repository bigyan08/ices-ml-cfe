# Speed comparision of operation on python list and numpy array

```python
import time

list1 = list(range(1000000))
sum = 0

# Using python list
start_time = time.time()
for num in list1:
  sum += num
end_time = time.time()

print(f"Sum of list1: {sum}")
print(f"Time taken on python list: {end_time-start_time}")
```

```python
np_array = np.array(range(1000000))

# Using numpy array
start_time = time.time()
sum = np.sum(np_array)
end_time = time.time()

print(f"Sum of np_array: {sum}")
print(f"Time taken on numpy array: {end_time-start_time}")
```

# NumPy Arrays Dimensions with Indexing

**1D Array Indexing**

```python
x = np.array([1, 2, 3, 4, 5])
print(x)
```

Index →   0  1  2  3  4
<br>
Value → 1  2  3  4  5

```python
print(x[0])   # first value
print(x[3])   # fourth value
print(x[-1])  # last value
```

**2D Indexing**

```python
class_marks = np.array(
          [[40, 50, 60],
          [70, 80, 90],
          [55, 65, 75]])

print(class_marks)
```

Row 0 → [40, 50, 60]
<br>
Row 1 → [70, 80, 90]
<br>
Row 2 → [55, 65, 75]
<br>
Indexing Format: **array[row_index][column_index]**

```python
print(class_marks[0][0])  # Row 0, Column 0 → 40
print(class_marks[1][2])  # Row 1, Column 2 → 90
print(class_marks[2][1])  # Row 2, Column 1 → ?
```

3D Indexing

```python
school_marks = np.array(
              [[[40, 50],
                [60, 70],
                [50, 60]],

              [[80, 85],
               [90, 95],
               [90, 100]]])

print(school_marks)
```

```python
school_marks.shape
```

Class 0
<br>
  Student 0 → [40, 50]
  <br>
  Student 1 → [60, 70]
  <br>
  Student 2 → [50, 60]

Class 1
<br>
  Student 0 → [80, 85]
  <br>
  Student 1 → [90, 95]
  <br>
  Student 2 → [90, 100]
<br>
Indexing: **array[class][row][column]**




```python
print(school_marks[0][0][1])  # Class 0, Student 0, Subject 1 → 50
print(school_marks[1][1][0])  # Class 1, Student 1, Subject 0 → 90
```

```python
print(school_marks[1, 2, 1])  # Class 1, Student 2, Subject 1 → 100
```

# **Randomness & Reproducibility**


```python
import random
```

```python
np.random.seed(1337) # reproducibility

np.random.randint(1,10) # Random integer from [1, 10)
```

**Different ways of creating random values**

**1) np.random.randint() → Random Integers**

```python
np.random.randint(1, 7) # single value: low (inclusive) → high (exclusive)
```

```python
np.random.randint(0, 100, size=5) # 1D array
```

```python
np.random.randint(10, 50, size=(2,3)) # 2D array
```

**2) np.random.rand() → Random Floats [0, 1)**
<br>
Used when:
1.   Probabilities
2.   ML weights



```python
np.random.rand()        # single value
```

```python
np.random.rand(5)       # 1D array
```

```python
np.random.rand(2, 3)    # 2D array
```

```python
np.random.random((2, 3)) # same as rand() but takes only one argument for 2D
```

**3) np.random.randn() → Random from Standard Normal Distribution**
<br>
Mean 0, Std Dev 1.
<br>
Values:

- Can be negative

- Mostly around 0

```python
np.random.randn()
```

```python
np.random.randn(5)
```

```python
np.random.randn(2, 3)
```

**4) np.random.shuffle() → Shuffle In-Place**

```python
arr = np.array([1,2,3,4,5])
print(arr)
np.random.shuffle(arr)
print(arr)
```

```python
# Decorators: enhance the base function
import time

def timer(base_fn):
  def enhanced_fn():
    start_time = time.time()
    base_fn()
    stop_time = time.time()
    print(f"Time taken: {stop_time - start_time} seconds")
  return enhanced_fn

@timer
def brewtea():
  print("Making Tea")
  time.sleep(1)
  print("Tea is ready")

brewtea()
```
