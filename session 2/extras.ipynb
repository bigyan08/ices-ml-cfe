# Lambda Function

A lambda function is a small anonymous function.

A lambda function can take any number of arguments, but can only have one expression.

syntax:
lambda arguments : expression

### single argument

```python
def increment(a):
  return a+1
```

```python
z = lambda a:a+1
```

```python
z(2)
```

### multiple arguments

```python
x = lambda a,b:a+b
```

```python
x(2,3)
```

## Mainly used with map(), filter(), sorted() functions

### map()
The function is used to map any iterables to the lambda function

```python
nums = [1,2,3,4]
squarred = list(map(lambda x:x**2, nums))
squarred
```

### filter()
This function is used to map any iterables only when the condition is True.

```python
nums = [1,2,3,4,5]
even = list(filter(lambda x: x%2==0,nums))
even
```

### sorted()
Uses lambda as key for custom sorting

```python
words = ["apple", "pie", "banana", "cherry"]
sorted_words = sorted(words, key=lambda x: len(x))
print(sorted_words)
```

```python
def average_grade(grades):
  if not grades:
    raise ValueError("empty grades not allowed")
  return sum(grades) / len(grades)
```

```python
average_grade([10,20])
```
